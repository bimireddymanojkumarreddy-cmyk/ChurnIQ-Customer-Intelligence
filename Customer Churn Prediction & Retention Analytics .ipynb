companies. Losing customers affects recurring revenue and
increases customer acquisition costs.

customers and support data-driven retention strategies.

### 1.3 Machine Learning Problem

Problem Type: Supervised Binary Classification

```python
# Create a copy for feature engineering
df_model = df_clean.copy()

# Average monthly charges per month of tenure
df_model["AverageChargesPerMonth"] = (
    df_model["TotalCharges"] / df_model["tenure"].replace(0, 1)
)

# Customer tenure groups
df_model["TenureGroup"] = pd.cut(
    df_model["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["New", "Short-term", "Medium-term", "Long-term"]
)

# Monthly charge categories
df_model["MonthlyChargeGroup"] = pd.cut(
    df_model["MonthlyCharges"],
    bins=[0, 35, 65, 100, float("inf")],
    labels=["Low", "Medium", "High", "Very High"],
    include_lowest=True
)

# Number of subscribed additional services
service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df_model["TotalAdditionalServices"] = (
    df_model[service_columns] == "Yes"
).sum(axis=1)

print("Feature engineering completed.")
print("New dataset shape:", df_model.shape)

display(
    df_model[
        [
            "tenure",
            "TotalCharges",
            "AverageChargesPerMonth",
            "TenureGroup",
            "MonthlyCharges",
            "MonthlyChargeGroup",
            "TotalAdditionalServices"
        ]
    ].head()
)
```

```python
from sklearn.model_selection import train_test_split

# Remove customerID because it is an identifier, not a useful predictor
X = df_model.drop(columns=["Churn", "customerID"])
y = df_model["Churn"].map({"No": 0, "Yes": 1})

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training shape:", X_train.shape)
print("Testing shape :", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify column types using training data only
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Preprocessing pipeline created successfully.")
```

```python
# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data without fitting again
X_test_processed = preprocessor.transform(X_test)

print("Training processed shape:", X_train_processed.shape)
print("Testing processed shape :", X_test_processed.shape)
print("Preprocessing completed successfully.")
```

```python
assert len(X_train) + len(X_test) == len(df_model)
assert X_train_processed.shape[0] == y_train.shape[0]
assert X_test_processed.shape[0] == y_test.shape[0]

print("All preprocessing checks passed.")
```

```python
from sklearn.linear_model import LogisticRegression

# Train baseline model
baseline_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

baseline_model.fit(X_train_processed, y_train)

print("Logistic Regression training completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Predictions
y_pred = baseline_model.predict(X_test_processed)
y_prob = baseline_model.predict_proba(X_test_processed)[:, 1]

# Evaluation metrics
print("BASELINE MODEL PERFORMANCE")
print("=" * 40)

print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1 Score :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob), 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Create validation data from training data
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train_processed,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Define models
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000, class_weight="balanced", random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(
        max_depth=6, class_weight="balanced", random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

# Train and evaluate
results = []

for name, model in models.items():
    model.fit(X_fit, y_fit)

    predictions = model.predict(X_val)
    probabilities = model.predict_proba(X_val)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_val, predictions),
        "Precision": precision_score(y_val, predictions, zero_division=0),
        "Recall": recall_score(y_val, predictions, zero_division=0),
        "F1 Score": f1_score(y_val, predictions, zero_division=0),
        "ROC-AUC": roc_auc_score(y_val, probabilities)
    })

# Display comparison
model_results = pd.DataFrame(results).set_index("Model")
display(model_results.round(3))
```

```python
import numpy as np
import tensorflow as tf

# Reproducibility
tf.keras.utils.set_random_seed(42)

# Convert sparse matrices if necessary
X_train_ann = (
    X_fit.toarray() if hasattr(X_fit, "toarray") else np.asarray(X_fit)
).astype("float32")

X_val_ann = (
    X_val.toarray() if hasattr(X_val, "toarray") else np.asarray(X_val)
).astype("float32")

y_train_ann = np.asarray(y_fit, dtype="float32")
y_val_ann = np.asarray(y_val, dtype="float32")

print("ANN Training Shape:", X_train_ann.shape)
print("ANN Validation Shape:", X_val_ann.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

ann_model = Sequential([
    Input(shape=(X_train_ann.shape[1],)),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(1, activation="sigmoid")
])

ann_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

ann_model.summary()
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = ann_model.fit(
    X_train_ann,
    y_train_ann,
    validation_data=(X_val_ann, y_val_ann),
    epochs=100,
    batch_size=32,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
ann_prob = ann_model.predict(X_val_ann).ravel()
ann_pred = (ann_prob >= 0.5).astype(int)

ann_results = {
    "Model": "ANN",
    "Accuracy": accuracy_score(y_val, ann_pred),
    "Precision": precision_score(y_val, ann_pred, zero_division=0),
    "Recall": recall_score(y_val, ann_pred, zero_division=0),
    "F1 Score": f1_score(y_val, ann_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_val, ann_prob)
}

display(pd.DataFrame([ann_results]).round(3))
```

```python
model_results = pd.concat([
    model_results,
    pd.DataFrame([ann_results]).set_index("Model")
])

display(model_results.round(3).sort_values("ROC-AUC", ascending=False))
```

```python
def recommend_retention_action(customer):
    """
    Generate a retention suggestion using customer characteristics.
    """

    recommendations = []

    contract = customer.get("Contract")
    tenure = customer.get("tenure", 0)
    monthly_charges = customer.get("MonthlyCharges", 0)
    security = customer.get("OnlineSecurity")
    support = customer.get("TechSupport")
    payment = customer.get("PaymentMethod")

    if contract == "Month-to-month":
        recommendations.append(
            "Offer a suitable long-term contract with a loyalty benefit."
        )

    if tenure <= 12:
        recommendations.append(
            "Provide an onboarding check-in and early customer support."
        )

    if monthly_charges >= 80:
        recommendations.append(
            "Review the customer's plan and offer a suitable cost-saving option."
        )

    if security == "No":
        recommendations.append(
            "Explain the benefits of online security and offer an appropriate plan."
        )

    if support == "No":
        recommendations.append(
            "Offer technical support assistance or a suitable support package."
        )

    if payment == "Electronic check":
        recommendations.append(
            "Offer convenient alternative payment methods and explain their benefits."
        )

    if not recommendations:
        recommendations.append(
            "Maintain regular engagement and monitor customer satisfaction."
        )

    return recommendations
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify feature types using only fitting data
numeric_features = X_fit_raw.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_fit_raw.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Numeric preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combined preprocessing
final_preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# Fit only on fitting data
X_fit_processed = final_preprocessor.fit_transform(X_fit_raw)

# Transform validation data without fitting again
X_val_processed = final_preprocessor.transform(X_val_raw)

print("Preprocessing completed successfully.")
print("Fitting data shape:", X_fit_processed.shape)
print("Validation data shape:", X_val_processed.shape)
```
